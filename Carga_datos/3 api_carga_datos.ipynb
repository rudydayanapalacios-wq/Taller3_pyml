### 1. Cargar los datos

```python
import requests
import pandas as pd

# Importar datos desde una API
url = "https://rickandmortyapi.com/api/character"

data = requests.get(url).json()

data_api = pd.DataFrame(data['results'])

data_api = data_api[['id', 'name', 'status', 'species', 'type', 'gender']]

print(data_api.head())
data_api.shape
data_api.isnull().sum()
```

### 2.Codificación de variables categóricas.

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

data_api['status_label'] = le.fit_transform(data_api['status'])

data_api[['status', 'status_label']].head()








from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False)

status_encoded = encoder.fit_transform(data_api[['status']])

pd.DataFrame(
    status_encoded,
    columns=encoder.get_feature_names_out(['status'])
).head()
```

### 3.Normalización y estandarización.

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Normalización
normalizador = MinMaxScaler()
data_api['id_normalizado'] = normalizador.fit_transform(data_api[['id']])

# Estandarización
estandarizador = StandardScaler()
data_api['id_estandarizado'] = estandarizador.fit_transform(data_api[['id']])

# Mostrar resultados
data_api[['id', 'id_normalizado', 'id_estandarizado']].head()
```

### 4.SMOTE

```python
from imblearn.over_sampling import SMOTE

# Separar características y variable objetivo
X = data_api[['id']]
y = data_api['status_label']

# Aplicar SMOTE
smote = SMOTE(random_state=42)
X_balanceado, y_balanceado = smote.fit_resample(X, y)

# Comprobar el balance
print("Antes:")
print(y.value_counts())

print("\nDespués:")
print(y_balanceado.value_counts())
```

```python
# Crear el conjunto de datos final
data_final = X_balanceado.copy()

# Agregar la variable objetivo
data_final['status'] = y_balanceado

# Mostrar las primeras filas
data_final.head()
```
