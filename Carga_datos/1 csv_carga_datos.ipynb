### 2. codifique variables categoricas usando LabelEncorder y One-Hot Encoding


```python
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder

# Label Encoding
le = LabelEncoder()
df['categoria_label'] = le.fit_transform(df['categoria'])

df[['categoria', 'categoria_label']].head()

# One-Hot Encoding
encoder = OneHotEncoder(sparse_output=False)
categoria_encoded = encoder.fit_transform(df[['categoria']])

pd.DataFrame(
    categoria_encoded,
    columns=encoder.get_feature_names_out(['categoria'])
).head()
```

### 3.Aplique normalización y estandarización


```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Normalización
normalizador = MinMaxScaler()
df['total_venta_normalizada'] = normalizador.fit_transform(df[['total_venta']])

# Estandarización
estandarizador = StandardScaler()
df['total_venta_estandarizada'] = estandarizador.fit_transform(df[['total_venta']])

# Mostrar resultados
df[['total_venta', 'total_venta_normalizada', 'total_venta_estandarizada']].head()
```

### 4.Balance la variable objetivo usando SMOTE.


```python
from imblearn.over_sampling import SMOTE

# Separar características y variable objetivo
X = df[['cantidad', 'precio_unitario', 'total_venta']]
y = df['categoria_label']

# Aplicar SMOTE
smote = SMOTE(random_state=42)
X_balanceado, y_balanceado = smote.fit_resample(X, y)

# Comprobar el balance
print("Antes:")
print(y.value_counts())

print("\nDespués:")
print(y_balanceado.value_counts())
```

### 5. Muestre el conjunto de datos al final listo para entender un modelo de clasicación

```python
# Crear el conjunto de datos final
df_final = X_balanceado.copy()

# Agregar la variable objetivo
df_final['categoria'] = y_balanceado

# Mostrar las primeras filas
df_final.head()
```
